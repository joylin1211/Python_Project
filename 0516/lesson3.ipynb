```python
#學生總分300
#學生可以加５％

try:
    scores = int(input("請輸入學生分數(最高300分):"))
except ValueError:
    print("輸入錯誤：請輸入整數格式的分數。")
else:
    is_add = input("符合加分條件(y,n)")
    if is_add == "y": #單向選擇
        scores = scores * 1.05
    scores = min(scores, 300)
    print("學生分數:", round(scores))
```

```python
try:
    score = int(input("請輸入成績(最高300)"))
except ValueError:
    print("輸入錯誤：請輸入整數格式的成績。")
else:
    a = input("是否要加分(y,n)")
    if a == "y":
        score = score * 1.05
    score = min(score, 300)
    print("學生的總分為:", round(score))
```
