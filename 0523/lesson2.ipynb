```python
import random
answer=random.randint(1,100)
count=0

print("猜數字遊戲")
print("請輸入一個0-100的數字")

while True:
    guess=int(input("輸入數字"))

    count+=1

    if guess>answer:
        print("太大了")
    elif guess<answer:
        print("太小了")
    else:
        print(f"恭喜猜對了！答案是 {answer}")
        print(f"你總共猜了 {count} 次")
        break
```
