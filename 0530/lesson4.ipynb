```python
import os
from pathlib import Path

cwd = Path.cwd()
candidate = cwd / "student.txt"
if not candidate.exists():
    candidate = cwd / "0530" / "student.txt"

print("Current working directory:", cwd)
print("Reading file:", candidate)

try:
    with candidate.open("r", encoding="utf-8") as file:
        data = file.read()
        print(data)
except FileNotFoundError:
    print("Error: student.txt not found. Check cwd or use 0530/student.txt")
except Exception as e:
    print("Error:", e)
```
