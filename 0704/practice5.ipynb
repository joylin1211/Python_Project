```python
import tkinter as tk
from tkinter import ttk, messagebox
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

# 設定 Matplotlib 中文字體（避免圖表中文變豆腐塊）
plt.rcParams["font.family"] = "Microsoft JhengHei" # Windows 使用微軟正黑體
plt.rcParams["axes.unicode_minus"] = False

# 1. 原始資料設定
tickers = {
    "台積電": "2330.TW",
    "聯電": "2303.TW",
    "聯發科": "2454.TW",
    "鴻海": "2317.TW",
}
code_to_name = {v: k for k, v in tickers.items()}

# 2. 建立主視窗
root = tk.Tk()
root.title("股票數據分析系統")
root.geometry("1000x700")  # ✅ 已修正：中間為小寫英文字母 x
root.configure(bg="#f5f6fa") # 使用現代感的淺灰色背景

# 調整 ttk 樣式
style = ttk.Style()
style.theme_use("clam")
style.configure("TButton", font=("Microsoft JhengHei", 11), padding=6)
style.configure("Main.TButton", background="#4a69bd", foreground="white", borderwidth=0)
style.map("Main.TButton", background=[("active", "#3b5398")])

# --- 側邊控制面板 ---
left_frame = tk.Frame(root, bg="#2f3640", width=250)
left_frame.pack(side=tk.LEFT, fill=tk.Y)
left_frame.pack_propagate(False)

title_label = tk.Label(left_frame, text="📊 股價分析工具", font=("Microsoft JhengHei", 16, "bold"), fg="white", bg="#2f3640")
title_label.pack(pady=30)

# 顯示目前監控的股票
stock_box = tk.LabelFrame(left_frame, text=" 監控目標 ", font=("Microsoft JhengHei", 10), fg="#dcdde1", bg="#2f3640", padx=10, pady=10)
stock_box.pack(pady=10, padx=15, fill=tk.X)

for name, code in tickers.items():
    lbl = tk.Label(stock_box, text=f"• {name} ({code})", font=("Microsoft JhengHei", 11), fg="#f5f6fa", bg="#2f3640", anchor="w")
    lbl.pack(fill=tk.X, pady=2)

# --- 右側主顯示區 ---
right_frame = tk.Frame(root, bg="#f5f6fa")
right_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True, padx=20, pady=20)

# 下方相關係數表格區
matrix_frame = tk.LabelFrame(right_frame, text=" 📊 日報酬率相關係數矩陣 ", font=("Microsoft JhengHei", 12, "bold"), bg="white", padx=15, pady=15)
matrix_frame.pack(side=tk.BOTTOM, fill=tk.X, pady=(10, 0))

# 上方圖表區
chart_frame = tk.LabelFrame(right_frame, text=" 📈 歷史收盤價走勢 ", font=("Microsoft JhengHei", 12, "bold"), bg="white")
chart_frame.pack(side=tk.TOP, fill=tk.BOTH, expand=True)

# 全域變數用來記錄畫布，方便重新整理
canvas = None

# --- 核心邏輯功能 ---
def run_analysis():
    global canvas
    
    # 顯示載入中提示
    status_lbl.config(text="資料下載與分析中，請稍候...", fg="#e1b12c")
    root.update()
    
    try:
        # 下載資料
        data = yf.download(
            list(tickers.values()),
            start="2026-01-01",
            interval="1d",
            auto_adjust=True
        )
        
        if data.empty:
            raise ValueError("未取得任何數據，請檢查網路或開市時間。")
            
        # 處理資料
        close = data['Close']
        close = close.rename(columns=code_to_name)
        
        returns = close.pct_change().dropna()
        corr = returns.corr()
        
        # 繪製圖表 (Matplotlib)
        fig, ax = plt.subplots(figsize=(7, 3.5), dpi=100)
        fig.patch.set_facecolor('white')
        
        for column in close.columns:
            ax.plot(close.index, close[column], marker='o', markersize=3, label=column, linewidth=1.5)
            
        ax.set_title("2026年起 收盤價走勢", fontsize=12, fontweight='bold', color='#2f3640')
        ax.set_xlabel("日期", fontsize=10)
        ax.set_ylabel("價格 (TWD)", fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.5)
        ax.legend(loc="upper left", frameon=True, facecolor='white', edgecolor='none')
        fig.tight_layout()
        
        # 清除舊圖表
        if canvas:
            canvas.get_tk_widget().destroy()
            
        # 將新圖表嵌入 Tkinter
        canvas = FigureCanvasTkAgg(fig, master=chart_frame)
        canvas.draw()
        canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True, padx=5, pady=5)
        
        # 渲染相關係數矩陣到表格 (Treeview)
        for widget in matrix_frame.winfo_children():
            widget.destroy() # 清空舊表格
            
        cols = ["股票名稱"] + list(corr.columns)
        tree = ttk.Treeview(matrix_frame, columns=cols, show="headings", height=4)
        tree.pack(fill=tk.X, expand=True)
        
        # 設定表頭
        for col in cols:
            tree.heading(col, text=col)
            tree.column(col, width=100, anchor="center")
            
        # 填入數據 (四捨五入到小數點後4位)
        for index, row in corr.iterrows():
            row_values = [index] + [f"{val:.4f}" for val in row]
            tree.insert("", tk.END, values=row_values)
            
        status_lbl.config(text="✅ 分析完成！", fg="#4cd137")
        
    except Exception as e:
        status_lbl.config(text="❌ 發生錯誤", fg="#c23616")
        messagebox.showerror("錯誤", f"執行失敗:\n{str(e)}")

# --- 側邊按鈕與狀態列 ---
btn_run = ttk.Button(left_frame, text="🚀 開始下載與分析", style="Main.TButton", command=run_analysis)
btn_run.pack(pady=20, padx=20, fill=tk.X)

status_lbl = tk.Label(left_frame, text="系統就緒", font=("Microsoft JhengHei", 10), fg="#ffffff", bg="#2f3640")
status_lbl.pack(side=tk.BOTTOM, pady=20)

# 啟動主程式
root.mainloop()
```

```python
import sys
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from matplotlib.backends.backend_qtagg import FigureCanvasQTAgg as FigureCanvas

from PySide6.QtWidgets import (QApplication, QMainWindow, QWidget, QHBoxLayout, 
                             QVBoxLayout, QPushButton, QLabel, QGroupBox, 
                             QTableWidget, QTableWidgetItem, QHeaderView, QMessageBox)
from PySide6.QtCore import Qt, QThread, Signal
from PySide6.QtGui import QFont

# 設定 Matplotlib 中文字體（避免圖表中文變豆腐塊）
plt.rcParams["font.family"] = "Microsoft JhengHei"
plt.rcParams["axes.unicode_minus"] = False

# 1. 股票資料設定
tickers = {
    "台積電": "2330.TW",
    "聯電": "2303.TW",
    "聯發科": "2454.TW",
    "鴻海": "2317.TW",
}
code_to_name = {v: k for k, v in tickers.items()}


# 2. 建立非同步下載線程（避免下載時造成 UI 卡死）
class WorkerThread(QThread):
    data_fetched = Signal(pd.DataFrame)
    error_occurred = Signal(str)

    def run(self):
        try:
            data = yf.download(
                list(tickers.values()),
                start="2026-01-01",
                interval="1d",
                auto_adjust=True
            )
            if data.empty:
                self.error_occurred.emit("未取得任何數據，請檢查網路或開市時間。")
            else:
                self.data_fetched.emit(data)
        except Exception as e:
            self.error_occurred.emit(str(e))


# 3. 主視窗設計
class StockAnalysisWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("股票高級數據分析系統 (PySide6 專業版)")
        self.resize(1100, 750)
        
        # 套用現代深色系 QSS 樣式表
        self.setStyleSheet("""
            QMainWindow { background-color: #1e272e; }
            QWidget { font-family: 'Microsoft JhengHei'; }
            
            /* 側邊欄樣式 */
            #SideBar { background-color: #2c3a47; border-right: 1px solid #3d4e5f; }
            #SideTitle { color: #ffffff; font-size: 18px; font-weight: bold; margin-bottom: 20px; }
            
            /* 群組外框樣式 */
            QGroupBox { 
                color: #ffffff; font-size: 13px; font-weight: bold;
                border: 1px solid #3d4e5f; border-radius: 6px; margin-top: 12px; padding-top: 12px;
            }
            QGroupBox::title { subcontrol-origin: margin; left: 10px; padding: 0 5px; }
            
            /* 按鈕樣式 */
            QPushButton {
                background-color: #00a8ff; color: white; font-size: 14px; font-weight: bold;
                border-radius: 5px; padding: 10px;
            }
            QPushButton:hover { background-color: #54a0ff; }
            QPushButton:pressed { background-color: #0086cc; }
            
            /* 標籤與狀態欄 */
            QLabel { color: #dcdde1; font-size: 13px; }
            #StatusLabel { font-size: 12px; font-weight: bold; }
            
            /* 表格樣式 */
            QTableWidget { 
                background-color: #2f3640; color: #ffffff; gridline-color: #3d4e5f; 
                border: 1px solid #3d4e5f; border-radius: 4px;
            }
            QHeaderView::section { 
                background-color: #1e272e; color: #00a8ff; font-weight: bold;
                padding: 5px; border: 1px solid #3d4e5f;
            }
        """)
        
        self.init_ui()

    def init_ui(self):
        # 主佈局（左右分割）
        main_widget = QWidget()
        self.setCentralWidget(main_widget)
        main_layout = QHBoxLayout(main_widget)
        main_layout.setContentsMargins(0, 0, 0, 0)
        main_layout.setSpacing(0)

        # ---------------- 左側控制面板 ----------------
        sidebar = QWidget()
        sidebar.setObjectName("SideBar")
        sidebar.setFixedWidth(260)
        sidebar_layout = QVBoxLayout(sidebar)
        sidebar_layout.setContentsMargins(20, 30, 20, 20)

        # 系統標題
        title_lbl = QLabel("📊 股價資產分析")
        title_lbl.setObjectName("SideTitle")
        title_lbl.setAlignment(Qt.AlignmentFlag.AlignCenter)
        sidebar_layout.addWidget(title_lbl)

        # 監控股票群組
        stock_box = QGroupBox(" 監控目標 ")
        stock_layout = QVBoxLayout(stock_box)
        for name, code in tickers.items():
            lbl = QLabel(f"• {name} ({code})")
            lbl.setStyleSheet("font-size: 14px; color: #f5f6fa; padding: 2px 0;")
            stock_layout.addWidget(lbl)
        sidebar_layout.addWidget(stock_box)
        
        sidebar_layout.addStretch()

        # 開始按鈕
        self.btn_run = QPushButton("🚀 開始下載與分析")
        self.btn_run.setCursor(Qt.CursorShape.PointingHandCursor)
        self.btn_run.clicked.connect(self.start_analysis)
        sidebar_layout.addWidget(self.btn_run)

        # 狀態標籤
        self.status_lbl = QLabel("系統就緒")
        self.status_lbl.setObjectName("StatusLabel")
        self.status_lbl.setAlignment(Qt.AlignmentFlag.AlignCenter)
        sidebar_layout.addWidget(self.status_lbl)

        main_layout.addWidget(sidebar)

        # ---------------- 右側主顯示區 ----------------
        right_container = QWidget()
        right_layout = QVBoxLayout(right_container)
        right_layout.setContentsMargins(20, 20, 20, 20)
        right_layout.setSpacing(15)

        # 上方圖表區
        self.chart_box = QGroupBox(" 📈 歷史收盤價走勢 ")
        self.chart_layout = QVBoxLayout(self.chart_box)
        self.chart_layout.setContentsMargins(10, 15, 10, 10)
        
        # 初始化預設提示文字
        self.placeholder_lbl = QLabel("請點擊左側按鈕開始加載分析數據")
        self.placeholder_lbl.setAlignment(Qt.AlignmentFlag.AlignCenter)
        self.placeholder_lbl.setStyleSheet("font-size: 14px; color: #718093;")
        self.chart_layout.addWidget(self.placeholder_lbl)
        right_layout.addWidget(self.chart_box, stretch=3)

        # 下方相關係數表格區
        self.table_box = QGroupBox(" 📊 日報酬率相關係數矩陣 ")
        self.table_layout = QVBoxLayout(self.table_box)
        self.table_layout.setContentsMargins(10, 15, 10, 10)

        self.table = QTableWidget()
        self.table_layout.addWidget(self.table)
        right_layout.addWidget(self.table_box, stretch=2)

        main_layout.addWidget(right_container)

    # ---------------- 核心邏輯功能 ----------------
    def start_analysis(self):
        self.btn_run.setEnabled(False)
        self.status_lbl.setText("⏳ 資料載入中，請稍候...")
        self.status_lbl.setStyleSheet("color: #f1c40f;")
        
        # 啟動線程
        self.worker = WorkerThread()
        self.worker.data_fetched.connect(self.handle_data)
        self.worker.error_occurred.connect(self.handle_error)
        self.worker.start()

    def handle_data(self, data):
        try:
            # 數據處理
            close = data['Close'].rename(columns=code_to_name)
            returns = close.pct_change().dropna()
            corr = returns.corr()

            # 1. 繪製 Matplotlib 圖表
            # 移除舊元件
            for i in reversed(range(self.chart_layout.count())): 
                self.chart_layout.itemAt(i).widget().setParent(None)

            fig, ax = plt.subplots(figsize=(6, 3), dpi=110)
            fig.patch.set_facecolor('#2f3640')  # 讓圖表背景融合右側區塊
            ax.set_facecolor('#2f3640')

            for column in close.columns:
                ax.plot(close.index, close[column], marker='o', markersize=3, label=column, linewidth=1.5)

            ax.set_title("2026年起 收盤價走勢趨勢", fontsize=11, color='white', fontweight='bold')
            ax.set_xlabel("日期", fontsize=9, color='#dcdde1')
            ax.set_ylabel("價格 (TWD)", fontsize=9, color='#dcdde1')
            
            # 設定座標軸顏色
            ax.tick_params(colors='#dcdde1', labelsize=8)
            ax.grid(True, linestyle='--', alpha=0.3, color='#718093')
            
            # 圖例樣式
            leg = ax.legend(loc="upper left", frameon=True, facecolor='#1e272e', edgecolor='none')
            for text in leg.get_texts():
                text.set_color('white')
                
            fig.tight_layout()

            # 將新畫布加入介面
            canvas = FigureCanvas(fig)
            self.chart_layout.addWidget(canvas)

            # 2. 渲染相關係數表格 (QTableWidget)
            self.table.setColumnCount(len(corr.columns) + 1)
            self.table.setRowCount(len(corr))
            
            headers = ["股票名稱"] + list(corr.columns)
            self.table.setHorizontalHeaderLabels(headers)
            
            # 自動拉伸填滿寬度
            self.table.horizontalHeader().setSectionResizeMode(QHeaderView.ResizeMode.Stretch)
            self.table.verticalHeader().setVisible(False)

            for row_idx, (index, row) in enumerate(corr.iterrows()):
                # 放第一欄的股票名稱
                name_item = QTableWidgetItem(index)
                name_item.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
                name_item.setForeground(Qt.globalColor.white)
                self.table.setItem(row_idx, 0, name_item)
                
                # 放相關係數數值
                for col_idx, val in enumerate(row):
                    val_item = QTableWidgetItem(f"{val:.4f}")
                    val_item.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
                    
                    # 對角線如果是 1.0000 標為金黃色亮點，其餘白色
                    if index == corr.columns[col_idx]:
                        val_item.setForeground(Qt.globalColor.yellow)
                    else:
                        val_item.setForeground(Qt.globalColor.white)
                        
                    self.table.setItem(row_idx, col_idx + 1, val_item)

            self.status_lbl.setText("✅ 分析完成！")
            self.status_lbl.setStyleSheet("color: #2ecc71;")

        except Exception as e:
            self.handle_error(str(e))
        finally:
            self.btn_run.setEnabled(True)

    def handle_error(self, err_msg):
        self.status_lbl.setText("❌ 發生錯誤")
        self.status_lbl.setStyleSheet("color: #e74c3c;")
        QMessageBox.critical(self, "錯誤提示", f"執行失敗:\n{err_msg}")
        self.btn_run.setEnabled(True)


if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = StockAnalysisWindow()
    window.show()
    sys.exit(app.exec())
```

```python
import sys
import requests
import pandas as pd
from PySide6.QtWidgets import (QApplication, QMainWindow, QWidget, QHBoxLayout, 
                             QVBoxLayout, QPushButton, QLabel, QTableWidget, 
                             QTableWidgetItem, QHeaderView, QLineEdit, QFrame, QMessageBox)
from PySide6.QtCore import Qt, QThread, Signal
from PySide6.QtGui import QFont, QColor

# 1. 非同步資料下載線程（直連證交所 API，快如閃電）
class TaiwanStockWorker(QThread):
    data_fetched = Signal(list)
    error_occurred = Signal(str)

    def run(self):
        try:
            # 台灣證券交易所：發行量加權股價指數成分股當日收盤行情
            url = "https://www.twse.com.tw/exchangeReport/MI_INDEX?response=json&type=ALL"
            response = requests.get(url, timeout=10)
            json_data = response.json()
            
            if json_data.get("stat") != "OK":
                self.error_occurred.emit("證交所伺服器回應異常，請稍後再試。")
                return

            # 證交所的股票資料通常在 tables[8] 或 tables[9]，我們動態尋找含有股票代號的表格
            raw_stocks = []
            for table in json_data.get("tables", []):
                if table.get("title") and "每日收盤行情" in table.get("title"):
                    # 欄位說明：0: 證券代號, 1: 證券名稱, 5: 收盤價, 9: 漲跌(+/-), 10: 漲跌價差
                    raw_stocks = table.get("data", [])
                    break
            
            if not raw_stocks:
                # 備用方案：尋找長度為 16 欄的數據（收盤行情的標準欄位數）
                for table in json_data.get("tables", []):
                    if table.get("data") and len(table["data"][0]) >= 11:
                        raw_stocks = table["data"]
                        break

            processed_stocks = []
            for item in raw_stocks:
                code = item[0].strip()
                name = item[1].strip()
                
                # 過濾掉權證、牛熊證等雜訊，只保留 4 位數標準台股股票
                if len(code) == 4 and code.isdigit():
                    try:
                        price = float(item[5].replace(",", "")) if item[5].strip() and item[5] != "--" else 0.0
                        
                        # 處理漲跌顏色判斷
                        sign = item[9]  # 可能是 綠色文字的 <p style='color:green'>-</p> 或直接是符号
                        change_val = float(item[10].replace(",", "")) if item[10].strip() and item[10] != "--" else 0.0
                        
                        if "-" in sign or "green" in sign:
                            change = -change_val
                        elif "+" in sign or "red" in sign:
                            change = change_val
                        else:
                            change = 0.0 if change_val == 0 else change_val
                            
                        # 計算粗略漲跌幅 (昨收 = 現價 - 漲跌)
                        prev_close = price - change
                        pct = (change / prev_close) * 100 if prev_close > 0 else 0.0
                        
                    except Exception:
                        price, change, pct = 0.0, 0.0, 0.0
                        
                    processed_stocks.append({
                        "code": code,
                        "name": name,
                        "price": price,
                        "change": change,
                        "pct": pct
                    })
                    
            self.data_fetched.emit(processed_stocks)
        except Exception as e:
            self.error_occurred.emit(str(e))


# 2. 主視窗設計（大數據美學儀表板）
class TaiwanStockDashboard(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("全台股票即時大數據監控儀表板")
        self.resize(1200, 850)
        self.all_stock_data = [] # 儲存原始資料以便搜尋
        
        # 頂級奢華科技感 QSS 樣式表
        self.setStyleSheet("""
            QMainWindow { background-color: #0d1117; }
            QWidget { font-family: 'Segoe UI', 'Microsoft JhengHei'; color: #c9d1d9; }
            
            /* 頂部控制列 */
            #HeaderZone { background-color: #161b22; border-bottom: 2px solid #21262d; }
            
            /* 搜尋框 */
            QLineEdit {
                background-color: #0d1117; color: #ffffff;
                border: 1px solid #30363d; border-radius: 20px;
                padding: 8px 15px; font-size: 14px;
            }
            QLineEdit:focus { border: 1px solid #58a6ff; }
            
            /* 按鈕 */
            QPushButton {
                background: qlineargradient(x1:0, y1:0, x2:1, y2:0, stop:0 #1f6feb, stop:1 #58a6ff);
                color: white; font-size: 14px; font-weight: bold;
                border-radius: 18px; padding: 8px 25px; border: none;
            }
            QPushButton:hover { background: qlineargradient(x1:0, y1:0, x2:1, y2:0, stop:0 #388bfd, stop:1 #79c0ff); }
            QPushButton:disabled { background: #21262d; color: #8b949e; }
            
            /* 數據卡片 */
            #Card { background-color: #161b22; border: 1px solid #30363d; border-radius: 10px; }
            
            /* 表格頂級優化 */
            QTableWidget { 
                background-color: #161b22; gridline-color: #21262d; 
                border: 1px solid #30363d; border-radius: 8px;
                font-size: 14px; alternate-background-color: #1c2128;
            }
            QTableWidget::item { padding: 12px; }
            QTableWidget::item:selected { background-color: #1f6feb; color: white; }
            
            QHeaderView::section { 
                background-color: #21262d; color: #58a6ff; font-weight: bold;
                padding: 10px; border: none; border-bottom: 2px solid #30363d; font-size: 14px;
            }
        """)
        
        self.init_ui()
        self.fetch_stock_data() # 開啟時自動讀取

    def init_ui(self):
        main_widget = QWidget()
        self.setCentralWidget(main_widget)
        main_layout = QVBoxLayout(main_widget)
        main_layout.setContentsMargins(25, 20, 25, 25)
        main_layout.setSpacing(18)

        # ---------------- 1. 頂部抬頭與控制列 ----------------
        header_widget = QWidget()
        header_widget.setObjectName("HeaderZone")
        header_layout = QHBoxLayout(header_widget)
        header_layout.setContentsMargins(15, 10, 15, 10)

        title_lbl = QLabel("📈 TAIWAN STOCK MARKET")
        title_lbl.setStyleSheet("font-size: 22px; font-weight: bold; color: #ffffff; letter-spacing: 1px;")
        header_layout.addWidget(title_lbl)
        
        header_layout.addStretch()

        # 搜尋輸入框
        self.search_input = QLineEdit()
        self.search_input.setPlaceholderText("🔍 輸入代碼或名稱快速篩選...")
        self.search_input.setFixedWidth(280)
        self.search_input.textChanged.connect(self.filter_data)
        header_layout.addWidget(self.search_input)

        # 重新整理按鈕
        self.btn_refresh = QPushButton("🔄 刷新行情")
        self.btn_refresh.clicked.connect(self.fetch_stock_data)
        header_layout.addWidget(self.btn_refresh)

        main_layout.addWidget(header_widget)

        # ---------------- 2. 數據綜合看板區 (Cards) ----------------
        cards_layout = QHBoxLayout()
        cards_layout.setSpacing(20)

        self.card_total = self.create_kpi_card("📊 監控總家數", "0", "#58a6ff")
        self.card_up = self.create_kpi_card("🔺 今日上漲", "0", "#39d353")
        self.card_down = self.create_kpi_card("🔻 今日下跌", "0", "#ff7b72")

        cards_layout.addWidget(self.card_total)
        cards_layout.addWidget(self.card_up)
        cards_layout.addWidget(self.card_down)
        main_layout.addLayout(cards_layout)

        # ---------------- 3. 中央數據大表格 ----------------
        self.table = QTableWidget()
        self.table.setAlternatingRowColors(True)
        self.table.setColumnCount(5)
        self.table.setHorizontalHeaderLabels(["股票代碼", "公司名稱", "當前股價 (TWD)", "昨日漲跌", "今日漲跌幅"])
        
        # 表格自適應與樣式細調
        self.table.horizontalHeader().setSectionResizeMode(QHeaderView.ResizeMode.Stretch)
        self.table.verticalHeader().setVisible(False)
        self.table.setEditTriggers(QTableWidget.EditTrigger.NoEditTriggers)
        self.table.setSelectionBehavior(QTableWidget.SelectionBehavior.SelectRows)

        main_layout.addWidget(self.table)

    # 快捷建立發光 KPI 卡片
    def create_kpi_card(self, title, val, color):
        card = QFrame()
        card.setObjectName("Card")
        layout = QVBoxLayout(card)
        layout.setContentsMargins(20, 15, 20, 15)
        
        t_lbl = QLabel(title)
        t_lbl.setStyleSheet("font-size: 13px; color: #8b949e; font-weight: bold;")
        v_lbl = QLabel(val)
        v_lbl.setStyleSheet(f"font-size: 28px; font-weight: bold; color: {color}; font-family: 'Impact';")
        
        layout.addWidget(t_lbl)
        layout.addWidget(v_lbl)
        card.setProperty("val_label", v_lbl) # 綁定指針方便後續動態修改數字
        return card

    # ---------------- 3. 核心大數據資料操作 ----------------
    def fetch_stock_data(self):
        self.btn_refresh.setEnabled(False)
        self.btn_refresh.setText("⚡ 讀取中...")
        
        self.worker = TaiwanStockWorker()
        self.worker.data_fetched.connect(self.display_data)
        self.worker.error_occurred.connect(self.handle_error)
        self.worker.start()

    def display_data(self, data_list):
        self.all_stock_data = data_list
        self.btn_refresh.setEnabled(True)
        self.btn_refresh.setText("🔄 刷新行情")
        
        # 更新上方看板數據
        total_count = len(data_list)
        up_count = sum(1 for x in data_list if x["change"] > 0)
        down_count = sum(1 for x in data_list if x["change"] < 0)
        
        self.card_total.property("val_label").setText(str(total_count))
        self.card_up.property("val_label").setText(f"{up_count} 家")
        self.card_down.property("val_label").setText(f"{down_count} 家")

        # 渲染表格
        self.render_table_rows(data_list)

    def render_table_rows(self, data):
        self.table.setRowCount(len(data))
        
        for row_idx, stock in enumerate(data):
            # 1. 代碼
            item_code = QTableWidgetItem(stock["code"])
            item_code.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
            item_code.setForeground(QColor("#8b949e"))
            
            # 2. 名稱
            item_name = QTableWidgetItem(stock["name"])
            item_name.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
            item_name.setFont(QFont("Microsoft JhengHei", 10, QFont.Weight.Bold))
            item_name.setForeground(QColor("#ffffff"))

            # 3. 價格
            item_price = QTableWidgetItem(f"{stock['price']:.2f}")
            item_price.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)
            item_price.setForeground(QColor("#58a6ff"))

            # 4. 漲跌價差與 5. 漲跌幅
            change_str = f"+{stock['change']:.2f}" if stock["change"] > 0 else f"{stock['change']:.2f}"
            pct_str = f"+{stock['pct']:.2f}%" if stock["pct"] > 0 else f"{stock['pct']:.2f}%"
            
            item_change = QTableWidgetItem(change_str)
            item_pct = QTableWidgetItem(pct_str)
            item_change.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)
            item_pct.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)

            # 根據正負號決定字體霓虹顏色（台灣習慣：漲紅、跌綠）
            if stock["change"] > 0:
                neon_color = QColor("#ff7b72") # 霓虹紅
                item_price.setFont(QFont("Segoe UI", 10, QFont.Weight.Bold))
            elif stock["change"] < 0:
                neon_color = QColor("#39d353") # 螢光綠
                item_price.setFont(QFont("Segoe UI", 10, QFont.Weight.Bold))
            else:
                neon_color = QColor("#c9d1d9") # 平盤白

            item_change.setForeground(neon_color)
            item_pct.setForeground(neon_color)

            # 填入表格
            self.table.setItem(row_idx, 0, item_code)
            self.table.setItem(row_idx, 1, item_name)
            self.table.setItem(row_idx, 2, item_price)
            self.table.setItem(row_idx, 3, item_change)
            self.table.setItem(row_idx, 4, item_pct)

    # 動態模糊過濾器
    def filter_data(self):
        text = self.search_input.text().strip().lower()
        if not text:
            self.render_table_rows(self.all_stock_data)
            return

        # 同步支援「代碼」或「名稱」關鍵字搜尋
        filtered = [x for x in self.all_stock_data if text in x["code"] or text in x["name"].lower()]
        self.render_table_rows(filtered)

    def handle_error(self, err_msg):
        self.btn_refresh.setEnabled(True)
        self.btn_refresh.setText("🔄 刷新行情")
        QMessageBox.critical(self, "連線錯誤", f"無法從台灣證券交易所抓取行情:\n{err_msg}")


if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = TaiwanStockDashboard()
    window.show()
    sys.exit(app.exec())
```

```python
import sys
import requests
from PySide6.QtWidgets import (QApplication, QMainWindow, QWidget, QHBoxLayout, 
                             QVBoxLayout, QPushButton, QLabel, QTableWidget, 
                             QTableWidgetItem, QHeaderView, QLineEdit, QFrame, QMessageBox)
from PySide6.QtCore import Qt, QThread, Signal
from PySide6.QtGui import QFont, QColor

# 1. 閃電資料抓取執行緒
class TaiwanStockWorker(QThread):
    data_fetched = Signal(list)
    error_occurred = Signal(str)

    def run(self):
        try:
            # 直連證交所全市場最新行情
            url = "https://www.twse.com.tw/exchangeReport/MI_INDEX?response=json&type=ALL"
            response = requests.get(url, timeout=10)
            json_data = response.json()
            
            if json_data.get("stat") != "OK":
                self.error_occurred.emit("證交所伺服器回應異常，請稍後再試。")
                return

            raw_stocks = []
            for table in json_data.get("tables", []):
                if table.get("title") and "每日收盤行情" in table.get("title"):
                    raw_stocks = table.get("data", [])
                    break
            
            if not raw_stocks:
                for table in json_data.get("tables", []):
                    if table.get("data") and len(table["data"][0]) >= 11:
                        raw_stocks = table["data"]
                        break

            processed_stocks = []
            for item in raw_stocks:
                code = item[0].strip()
                name = item[1].strip()
                
                # 只保留標準 4 位數台股
                if len(code) == 4 and code.isdigit():
                    try:
                        price = float(item[5].replace(",", "")) if item[5].strip() and item[5] != "--" else 0.0
                        sign = item[9]  
                        change_val = float(item[10].replace(",", "")) if item[10].strip() and item[10] != "--" else 0.0
                        
                        if "-" in sign or "green" in sign:
                            change = -change_val
                        elif "+" in sign or "red" in sign:
                            change = change_val
                        else:
                            change = 0.0 if change_val == 0 else change_val
                            
                        prev_close = price - change
                        pct = (change / prev_close) * 100 if prev_close > 0 else 0.0
                    except Exception:
                        price, change, pct = 0.0, 0.0, 0.0
                        
                    processed_stocks.append({
                        "code": code, "name": name, "price": price, "change": change, "pct": pct
                    })
            self.data_fetched.emit(processed_stocks)
        except Exception as e:
            self.error_occurred.emit(str(e))

# 2. 奢華科技風主介面
class TaiwanStockDashboard(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("全台股票即時大數據監控儀表板")
        self.resize(1100, 800)
        self.all_stock_data = [] 
        
        # 霓虹暗黑 Bloomberg 風格樣式表
        self.setStyleSheet("""
            QMainWindow { background-color: #0d1117; }
            QWidget { font-family: 'Segoe UI', 'Microsoft JhengHei'; color: #c9d1d9; }
            QLineEdit {
                background-color: #0d1117; color: #ffffff; border: 1px solid #30363d;
                border-radius: 18px; padding: 6px 15px; font-size: 13px;
            }
            QLineEdit:focus { border: 1px solid #58a6ff; }
            QPushButton {
                background: qlineargradient(x1:0, y1:0, x2:1, y2:0, stop:0 #1f6feb, stop:1 #58a6ff);
                color: white; font-weight: bold; border-radius: 15px; padding: 6px 20px;
            }
            QPushButton:hover { background: qlineargradient(x1:0, y1:0, x2:1, y2:0, stop:0 #388bfd, stop:1 #79c0ff); }
            QPushButton:disabled { background: #21262d; color: #8b949e; }
            #Card { background-color: #161b22; border: 1px solid #30363d; border-radius: 8px; }
            QTableWidget { 
                background-color: #161b22; gridline-color: #21262d; 
                border: 1px solid #30363d; border-radius: 8px; font-size: 14px;
                alternate-background-color: #1c2128;
            }
            QTableWidget::item { padding: 10px; }
            QTableWidget::item:selected { background-color: #1f6feb; color: white; }
            QHeaderView::section { 
                background-color: #21262d; color: #58a6ff; font-weight: bold;
                padding: 8px; border: none; border-bottom: 2px solid #30363d;
            }
        """)
        self.init_ui()
        self.fetch_stock_data()

    def init_ui(self):
        main_widget = QWidget()
        self.setCentralWidget(main_widget)
        main_layout = QVBoxLayout(main_widget)
        main_layout.setContentsMargins(20, 15, 20, 20)
        main_layout.setSpacing(15)

        # 頂部控制列
        header_layout = QHBoxLayout()
        title_lbl = QLabel("📈 TAIWAN STOCK MARKET")
        title_lbl.setStyleSheet("font-size: 20px; font-weight: bold; color: #ffffff;")
        header_layout.addWidget(title_lbl)
        header_layout.addStretch()

        self.search_input = QLineEdit()
        self.search_input.setPlaceholderText("🔍 輸入代碼或名稱快速篩選...")
        self.search_input.setFixedWidth(250)
        self.search_input.textChanged.connect(self.filter_data)
        header_layout.addWidget(self.search_input)

        self.btn_refresh = QPushButton("🔄 刷新行情")
        self.btn_refresh.clicked.connect(self.fetch_stock_data)
        header_layout.addWidget(self.btn_refresh)
        main_layout.addLayout(header_layout)

        # 看板區
        cards_layout = QHBoxLayout()
        self.card_total = self.create_kpi_card("📊 監控總家數", "0", "#58a6ff")
        self.card_up = self.create_kpi_card("🔺 今日上漲", "0", "#ff7b72")   # 台灣習慣：上漲紅
        self.card_down = self.create_kpi_card("🔻 今日下跌", "0", "#39d353") # 台灣習慣：下跌綠
        cards_layout.addWidget(self.card_total)
        cards_layout.addWidget(self.card_up)
        cards_layout.addWidget(self.card_down)
        main_layout.addLayout(cards_layout)

        # 表格區
        self.table = QTableWidget()
        self.table.setAlternatingRowColors(True)
        self.table.setColumnCount(5)
        self.table.setHorizontalHeaderLabels(["股票代碼", "公司名稱", "當前股價 (TWD)", "昨日漲跌", "今日漲跌幅"])
        self.table.horizontalHeader().setSectionResizeMode(QHeaderView.ResizeMode.Stretch)
        self.table.verticalHeader().setVisible(False)
        self.table.setEditTriggers(QTableWidget.EditTrigger.NoEditTriggers)
        self.table.setSelectionBehavior(QTableWidget.SelectionBehavior.SelectRows)
        main_layout.addWidget(self.table)

    def create_kpi_card(self, title, val, color):
        card = QFrame()
        card.setObjectName("Card")
        layout = QVBoxLayout(card)
        layout.setContentsMargins(15, 12, 15, 12)
        t_lbl = QLabel(title)
        t_lbl.setStyleSheet("font-size: 12px; color: #8b949e; font-weight: bold;")
        v_lbl = QLabel(val)
        v_lbl.setStyleSheet(f"font-size: 26px; font-weight: bold; color: {color}; font-family: 'Impact';")
        layout.addWidget(t_lbl)
        layout.addWidget(v_lbl)
        card.setProperty("val_label", v_lbl) 
        return card

    def fetch_stock_data(self):
        self.btn_refresh.setEnabled(False)
        self.btn_refresh.setText("⚡ 讀取中...")
        self.worker = TaiwanStockWorker()
        self.worker.data_fetched.connect(self.display_data)
        self.worker.error_occurred.connect(self.handle_error)
        self.worker.start()

    def display_data(self, data_list):
        self.all_stock_data = data_list
        self.btn_refresh.setEnabled(True)
        self.btn_refresh.setText("🔄 刷新行情")
        
        total_count = len(data_list)
        up_count = sum(1 for x in data_list if x["change"] > 0)
        down_count = sum(1 for x in data_list if x["change"] < 0)
        
        self.card_total.property("val_label").setText(str(total_count))
        self.card_up.property("val_label").setText(f"{up_count} 家")
        self.card_down.property("val_label").setText(f"{down_count} 家")
        self.render_table_rows(data_list)

    def render_table_rows(self, data):
        self.table.setRowCount(len(data))
        for row_idx, stock in enumerate(data):
            item_code = QTableWidgetItem(stock["code"])
            item_code.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
            item_code.setForeground(QColor("#8b949e"))
            
            item_name = QTableWidgetItem(stock["name"])
            item_name.setTextAlignment(Qt.AlignmentFlag.AlignCenter)
            item_name.setFont(QFont("Microsoft JhengHei", 10, QFont.Weight.Bold))
            item_name.setForeground(QColor("#ffffff"))

            item_price = QTableWidgetItem(f"{stock['price']:.2f}")
            item_price.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)
            item_price.setForeground(QColor("#58a6ff"))

            c_str = f"+{stock['change']:.2f}" if stock["change"] > 0 else f"{stock['change']:.2f}"
            p_str = f"+{stock['pct']:.2f}%" if stock["pct"] > 0 else f"{stock['pct']:.2f}%"
            
            item_change = QTableWidgetItem(c_str)
            item_pct = QTableWidgetItem(p_str)
            item_change.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)
            item_pct.setTextAlignment(Qt.AlignmentFlag.AlignRight | Qt.AlignmentFlag.AlignVCenter)

            if stock["change"] > 0:
                neon_color = QColor("#ff7b72")
                item_price.setFont(QFont("Segoe UI", 10, QFont.Weight.Bold))
            elif stock["change"] < 0:
                neon_color = QColor("#39d353")
                item_price.setFont(QFont("Segoe UI", 10, QFont.Weight.Bold))
            else:
                neon_color = QColor("#c9d1d9")

            item_change.setForeground(neon_color)
            item_pct.setForeground(neon_color)

            self.table.setItem(row_idx, 0, item_code)
            self.table.setItem(row_idx, 1, item_name)
            self.table.setItem(row_idx, 2, item_price)
            self.table.setItem(row_idx, 3, item_change)
            self.table.setItem(row_idx, 4, item_pct)

    def filter_data(self):
        text = self.search_input.text().strip().lower()
        if not text:
            self.render_table_rows(self.all_stock_data)
            return
        filtered = [x for x in self.all_stock_data if text in x["code"] or text in x["name"].lower()]
        self.render_table_rows(filtered)

    def handle_error(self, err_msg):
        self.btn_refresh.setEnabled(True)
        self.btn_refresh.setText("🔄 刷新行情")
        QMessageBox.critical(self, "連線錯誤", f"無法從證交所抓取行情:\n{err_msg}")

# 3. Jupyter Notebook 專用安全啟動進入點
if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    
    try:
        window.close()
    except NameError:
        pass
        
    window = TaiwanStockDashboard()
    window.show()
    window.raise_()
    window.activateWindow()
```
